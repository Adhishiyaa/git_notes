```python
"""Workshop Participant Greeting application (Python + Tkinter)."""

import tkinter as tk

WORKSHOP_DATE = "November 15, 2026"


def check_in():
    """Read the typed name and update the output Text widget."""
    name = name_entry.get().strip()

    output_text.config(state="normal")   # unlock so we can edit
    output_text.delete("1.0", tk.END)    # clear old message

    if not name:
        message = "Please enter your name before checking in."
    else:
        message = (
            f"Welcome, {name}!\n"
            f"You are checked in for the workshop on {WORKSHOP_DATE}.\n"
            "We're glad to have you with us. Enjoy the session!"
        )

    output_text.insert(tk.END, message)
    output_text.config(state="disabled")  # lock so the user can't type in it


# --- Main window ---
root = tk.Tk()
root.title("Workshop Participant Greeting")
root.geometry("420x320")
root.resizable(False, False)

# --- Instruction labels ---
title_label = tk.Label(root, text="Workshop Check-In", font=("Arial", 16, "bold"))
title_label.pack(pady=(15, 5))

instruction_label = tk.Label(
    root, text="Type your name below, then click the Check In button."
)
instruction_label.pack(pady=(0, 10))

name_label = tk.Label(root, text="Participant name:")
name_label.pack()

# --- Entry for the participant's name ---
name_entry = tk.Entry(root, width=30)
name_entry.pack(pady=5)
name_entry.focus()

# --- Check In button ---
check_in_button = tk.Button(root, text="Check In", command=check_in)
check_in_button.pack(pady=10)

# --- Multi-line output area ---
output_text = tk.Text(root, height=6, width=48, wrap="word", state="disabled")
output_text.pack(padx=10, pady=5)

# Pressing Enter in the Entry also checks in
root.bind("<Return>", lambda event: check_in())

root.mainloop()
```
